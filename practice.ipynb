```python
# Swapping Program without using third variable
a=10
b=12
print("Before swapping")
print("a:",a)
print("b:",b)
a,b=b,a
print("After swapping")
print("a:",a)
print("b:",b)
```

```python
# Swapping Program using third variable
a=int(input("enter the value of a:"))
b=int(input("enter the value of b:"))
print("Before swapping")
print("a:",a)
print("b:",b)
c=a
a=b
b=c
print("After swapping")
print("a:",a)
print("b:",b)
```
