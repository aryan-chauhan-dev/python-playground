```python
jackpot=random.randint(1,100)
guess=int(input("Ready to guess:"))
counter=1
while guess!=jackpot:
    if guess<jackpot:
        print("guess higher")
    else:
        print("Guess lower")
    guess=int(input("Ready to guess"))
    counter+=1
print("Right answer")
print("You took",counter,"attempts")
```
